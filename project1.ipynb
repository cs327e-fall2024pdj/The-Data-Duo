In our SQL Query we used the WHERE clause to only display customer's last names who begin with a Ca, and ordered them alphabetically by last name.

```python
!mysql sakila -e "SELECT customer_id, first_name, last_name FROM customer WHERE last_name LIKE 'ca%' ORDER BY last_name"
```

This SQL statement inserts a new row into the actor table with an actor named Glen McConaughey, an actor ID of 240, and the last update in 2006.

```python
!mysql sakila -e "INSERT INTO actor VALUES (240,'GLEN','McConaughey','2006-02-15 04:34:33')"
```

In this SQL statement, we updated the last_update timestamp of the actor with ID 1 to the current date and time.

```python
!mysql sakila -e "UPDATE actor SET last_update = '2024-09-25 06:07:33' WHERE actor_id = 1"
```

```python
!mysql sakila -e "SELECT actor_id, first_name, last_update FROM actor WHERE actor_id = 1" #To show that it was updated
```
