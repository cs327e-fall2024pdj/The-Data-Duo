```python
gsutil_cmd = "gsutil ls gs://cs327e-open-access-2/school-enrollments/*"
file_listings = os.popen(gsutil_cmd)
dataset="school_enrollments"

for file in file_listings:

    start_index = file.rindex("_") + 1
    end_index = file.rindex(".")
    table = file[start_index:end_index]
    
    if table in ("co", "district", "fl", "ga", "id", "ma", "mo", "ne", "ny", "or", "school", "sd", "wi"):
        print("skipping " + table)
        continue

    bq_cmd = "bq --location=US load --autodetect --skip_leading_rows=1 "\
             "--source_format=CSV " + dataset + "." + table + ' ' + file
    print(bq_cmd)
    
    os.system(bq_cmd)
```

Finds the number of different types of schools in Dallas County and orders the results by the count.

```python
%%bigquery
SELECT COUNT(ccd_school_type) AS Count, ccd_school_type AS school_type 
FROM school_enrollments.tx 
WHERE county = 'DALLAS COUNTY' 
GROUP BY ccd_school_type 
ORDER BY Count;
```

Calculates the average enrollment for each school in the Texas dataset, filtering out null values and including only schools with an average enrollment greater than 1000, and orders the results by average enrollment.

```python
%%bigquery
SELECT school, school_nces_id AS school_id, AVG(SAFE_CAST(REPLACE(REPLACE(REPLACE(total, ',', ''), '<', ''), '>', '') AS FLOAT64)) AS enrollment_average 
FROM school_enrollments.tx WHERE SAFE_CAST(REPLACE(REPLACE(REPLACE(total, ',', ''), '<', ''), '>', '') AS FLOAT64) IS NOT NULL 
GROUP BY school_nces_id, school 
HAVING AVG(SAFE_CAST(REPLACE(REPLACE(REPLACE(total, ',', ''), '<', ''), '>', '') AS FLOAT64)) > 1000
ORDER BY enrollment_average;
```

Counts the number of charter schools for each district in texas and filters the results to include where it says TRUE for charter school, ordering the results by the number of charter schools.

```python
%%bigquery
SELECT district_nces_id AS District_ID, count(ccd_charter_school) AS Num_charter_schools 
FROM school_enrollments.tx 
WHERE ccd_charter_school IS TRUE 
GROUP BY district_nces_id 
ORDER BY count(ccd_charter_school); 
```

Finds the count of high schools in each district in Texas by filtering out schools where their lowest grade is 8 or less, and orders the results by the high school count.

```python
%%bigquery
SELECT count(ccd_school_low_grade) AS Highschool_count, district_nces_id AS District_ID 
FROM school_enrollments.tx 
WHERE SAFE_CAST(REPLACE(REPLACE(ccd_school_low_grade, ",", ""), "<", "") AS FLOAT64) > 8 
GROUP BY district_nces_id
ORDER BY count(ccd_school_low_grade);
```

Calculates the average number of females in two different datasets for Wyoming and South Carolina, grouping the results by district, including only those districts with an average greater than 500, and combines the results ordering by the average number of females.

```python
%%bigquery
SELECT AVG(SAFE_CAST(female AS FLOAT64)) AS average_female, district AS District 
FROM school_enrollments.wy 
GROUP BY district HAVING AVG(SAFE_CAST(female AS FLOAT64))> 500 
UNION ALL 
SELECT AVG(SAFE_CAST(female AS FLOAT64)) AS average_female, district AS District 
FROM school_enrollments.sc 
GROUP BY district 
HAVING AVG(SAFE_CAST(female AS FLOAT64))> 500 
ORDER BY average_female;
```

Calculates the total enrollement for each district in Maryland, filtering to include only districts with a total enrollment count greater than 100,000, and orders the results by the total sum.

```python
%%bigquery
SELECT SUM(SAFE_CAST(REPLACE(REPLACE(total, ',', ''), '<', '') AS FLOAT64)) AS total_sum, district 
FROM school_enrollments.md 
GROUP BY district 
HAVING SUM(SAFE_CAST(REPLACE(REPLACE(total, ',', ''), '<', '') AS FLOAT64)) >  100000 
ORDER BY total_sum
```

Calculates the total number of students receiving free and reduced lunch for each district in Minnesota, filtering to include only districts with a total number of kids enrolled greater than 150000, and orders the results by the total.

```python
%%bigquery
SELECT SUM(SAFE_CAST(REPLACE(REPLACE(free_and_reduced, ',', ''), '<', '') AS FLOAT64)) AS Reduced_Lunch, district 
FROM school_enrollments.mn 
GROUP BY district 
HAVING SUM(SAFE_CAST(REPLACE(REPLACE(free_and_reduced, ',', ''), '<', '') AS FLOAT64)) > 150000 
ORDER BY Reduced_Lunch
```

This query calculates total disabilities by district in both Maryland and Minnesota, filtering to include only districts with a total enrollement greater than 20,000, and combines the results, ordering by the total number of disabilities.

```python
%%bigquery
SELECT district, 
       SUM(SAFE_CAST(REPLACE(REPLACE(disability, ',', ''), '<', '') AS FLOAT64)) AS total_disabilities 
FROM school_enrollments.md 
WHERE SAFE_CAST(REPLACE(REPLACE(total, ',', ''), '<', '') AS FLOAT64) > 20000 AND SAFE_CAST(REPLACE(REPLACE(disability, ',', ''), '<', '') AS FLOAT64) IS NOT NULL
GROUP BY district 

UNION ALL 

SELECT district, 
       SUM(SAFE_CAST(REPLACE(REPLACE(disability, ',', ''), '<', '') AS FLOAT64)) AS total_disabilities 
FROM school_enrollments.mn 
WHERE total > 20000 AND SAFE_CAST(REPLACE(REPLACE(disability, ',', ''), '<', '') AS FLOAT64) IS NOT NULL
GROUP BY district 

ORDER BY total_disabilities ASC;
```
