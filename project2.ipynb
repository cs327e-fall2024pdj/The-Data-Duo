In this query, we are retrieving the IDs of the films that fall under the Comedy category. By using an inner join between the film_category and category tables, we filter for films categorized as comedy and display them in descending order based on their film IDs.

```python
!psql -d postgres -c "SELECT f.film_id, c.category_id FROM pagila.film_category AS f INNER JOIN pagila.category AS c ON c.category_id = f.category_id WHERE c.name = 'Comedy' ORDER BY f.film_id DESC LIMIT 10;"
```

In this query, we select payment records with a payment amount greater than 5, retrieve the first and last name, customer ID, and payment information by joining the customer and payment tables, and sort them in descending order based on the payment amount to see the largest payments made.

```python
!psql -d postgres -c "SELECT p.payment_id, p.amount, c.first_name, c.last_name, c.customer_id FROM pagila.customer c INNER JOIN pagila.payment p ON c.customer_id = p.customer_id WHERE p.amount > 5 ORDER BY p.amount DESC LIMIT 15;"
```

In this query, we are retrieving a list of film titles, their language, and their lengths for the films that are in English by using the WHERE statement. We then order the results by the film’s length and limit it to the 15 shortest films.

```python
!psql -d postgres -c "SELECT f.title, l.name, f.length FROM pagila.film as f INNER JOIN pagila.language as l ON f.language_id = l.language_id WHERE l.name = 'English' ORDER BY f.length LIMIT 15;"
```

In this query, we select payment transactions with less than 3 and sort them by payment date in descending order to see the most recent transaction. By using a full join between the customer and payment tables, we obtain all of the relevant customer and payment data, even if some of the customers haven't made a payment or if some of the payments don't have customer details.

```python
!psql -d postgres -c "SELECT c.customer_id, c.first_name, p.payment_id, p.amount, p.payment_date FROM pagila.customer c FULL JOIN pagila.payment p ON c.customer_id = p.customer_id WHERE p.amount < 3 ORDER BY p.payment_date DESC LIMIT 10;"
```

In this query, we pull all of the city names that start with an 'A' along with their country names, and the last update time, sorted by country and limited to 10 results. The join is used to make sure we include all cities and countries, even if some don't have a matching pair.

```python
!psql -d postgres -c "SELECT ci.city, co.country, ci.last_update FROM pagila.city AS ci FULL JOIN pagila.country AS co ON co.country_id = ci.country_id WHERE ci.city Like 'A%' ORDER BY co.country LIMIT 10;" 
```

In this query, we retrieve the staff members with passwords that start with an '8', along with their address details, and order them by their last name. The join we use ensures that we include all of the staff, even if some of them don’t have an address listed in the table.

```python
!psql -d postgres -c "SELECT s.first_name, s.last_name, s.password, a.address_id, a.district FROM pagila.staff AS s LEFT JOIN pagila.address AS a ON s.address_id=a.address_id WHERE s.password LIKE '8%' ORDER BY s.last_name;" 
```

This query uses a self join to retrieve pairs of films that share the same length, displaying their titles and the length, while ensuring that each pair is listed only once and limited to 13 results.

```python
!psql -d postgres -c "SELECT f1.title AS film_1, f2.title AS film_2, f1.length FROM pagila.film f1 JOIN pagila.film f2 ON f1.length = f2.length WHERE f1.film_id < f2.film_id LIMIT 13;"
```
